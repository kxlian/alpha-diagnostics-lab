# Alpha Diagnostics Lab

This notebook walks through the core research workflow: simulate alpha scores, evaluate Rank IC, test information decay, and inspect long-short quantile behavior.

## 1. Simulate synthetic alpha research data

The data is synthetic by design. The goal is to isolate the diagnostic process rather than claim a live market edge.

```python
from alpha_diagnostics.simulate import simulate_alpha_dataset

dataset = simulate_alpha_dataset()
returns = dataset.returns
alpha_scores = dataset.alpha_scores
returns.head()
```

## 2. Rank IC

Rank IC tests whether higher-ranked securities by signal tend to achieve higher future returns.

```python
from alpha_diagnostics.diagnostics import compute_rank_ic

rank_ic = compute_rank_ic(alpha_scores, returns, horizon=1)
rank_ic.describe()
```

## 3. Information Decay

A signal should be tested across different horizons to understand how quickly its predictive relationship fades.

```python
from alpha_diagnostics.diagnostics import compute_information_decay

information_decay = compute_information_decay(alpha_scores, returns)
information_decay
```

## 4. Long-short quantile spread

Sorting assets into quantiles provides an intuitive economic interpretation of the signal.

```python
from alpha_diagnostics.diagnostics import compute_quantile_returns, compute_quantile_spread

quantile_returns = compute_quantile_returns(alpha_scores, returns)
quantile_spread = compute_quantile_spread(quantile_returns)
quantile_spread.tail()
```

## 5. Interpretation

A positive Rank IC, constructive decay profile, and upward-sloping quantile spread together provide preliminary evidence that the synthetic signal contains usable cross-sectional information.
